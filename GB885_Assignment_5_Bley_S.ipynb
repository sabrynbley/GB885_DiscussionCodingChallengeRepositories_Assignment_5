### **Dataset Design & Creation:**

The Strava Running Activity Data dataset was published on Kaggle by Ajit Jadhav and contains personal running activity data collected between March 2022 and December 2023. The data was exported using the Strava API, which records information from GPS-enabled fitness activities.

This dataset includes information about individual running activities, such as the activity date, distance, elapsed time, moving time, average pace or speed, elevation gain, heart rate (when available), and other workout metrics.

```python
# Download latest path to the data in Kaggle
path = kagglehub.dataset_download("ajitjadhav1/strava-running-activity-data")

# Get the data file name
file_name = os.listdir(path)[0]

# Define the full path to the data Excel
file_path = os.path.join(path, file_name)

# Load the data into a dataframe
running_df = pd.read_excel(file_path)
```

### **Data Cleaning:**

**Fix Column Data Types:**


The date columns were imported as strings, so convert them to be datetime objects. This will allow for easier use in later analysis.

```python
# Convert to datetime objects
running_df['start_date'] = pd.to_datetime(running_df['start_date'])
running_df['start_date_local'] = pd.to_datetime(running_df['start_date_local'])
```

**Remove Unneeded Activites:**

Out of 105 records, only 11 of the activities are not runs. We will focus on the running activites, so we can remove the other activity records.

```python
# Remove walks, workouts, and hikes
running_df = running_df[~running_df['type'].isin(['Walk', 'Hike', 'Workout'])]
```

**Remove Missing Lat & Lon Pairs:**

There were initially 9 records missing (there are not NA values, but they are empty pairs) a start and end latitude and longitude coordinates. For consistency in quality, we will remove the records that are missing this.

```python
running_df = running_df[
    (running_df['start_latlng'] != '[]') &
    (running_df['end_latlng'] != '[]')
]
```

**Remove Unneeded Columns:**

The sports type and type columns are essentially duplicates of each other. We only need one of them. Sr. no. is just what number in the table the record is. It doesn't hold any contextual value. Additionally, there are metrics for the activity's post on Strava's platform. It doesn't provide value for analysis of the activity itself.

```python
running_df = running_df.drop(columns=['sport_type', 'Sr. no.', 'achievement_count', 'kudos_count', 'comment_count', 'athlete_count', 'photo_count'])
```

**Rename Columns:**

The Kaggle dataset page (https://www.kaggle.com/datasets/ajitjadhav1/strava-running-activity-data) has some information about the units the values are measured in. To better keep track of that information, we can include it in the column names.

```python
running_df.rename(columns={'distance': 'distance_in_meters',
                           'moving_time': 'moving_time_in_secs',
                           'elapsed_time': 'elapsed_time_in_secs',
                           'total_elevation_gain': 'total_elevation_gain_in_meters',
                           'start_date': 'start_date_utc',
                           'average_speed': 'average_speed_in_mps',
                           'max_speed': 'max_speed_in_mps'}, inplace=True)
```
